```python
#Outlier Detection & Handling
negative_sales=df[df['Sales Price']<0]
df=df[df['Sales Price']>=0]
print(f"Rows Removed : {negative_sales.shape[0]}")
```

```python
def remove_outliers(df, column):
    Q1=df[column].quantile(0.25)
    Q3=df[column].quantile(0.75)
    IQR=Q3-Q1
    lower_bound=Q1-(3*IQR)
    upper_bound=Q1+(3*IQR)

    outliers=df[(df[column]< lower_bound) | (df[column]> upper_bound)]
    print(f"Outliers found: {outliers.shape[0]} rows")

    df_cleaned=df[~((df[column]< lower_bound) | (df[column]> upper_bound))]

    return df_cleaned

cleaned_df=remove_outliers(df,'Sales Price')
cleaned_df
```

```python
#Customer Segmentation & Analysis
sales_summary=df.groupby('Customer ID')['Total Sales'].transform('sum')
profit_summary=df.groupby('Customer ID')['Total Profit'].transform('sum')

df.loc[:, 'Sales_Quantile'] = (
    pd.qcut(sales_summary.rank(method='first'),5, labels=False, duplicates='drop') +1
)

df.loc[:, 'Profit_Quantile'] = (
    pd.qcut(profit_summary.rank(method='first'),5, labels=False, duplicates ='drop') +1
)

cross_grid = pd.crosstab(
    df['Sales_Quantile'],df['Profit_Quantile']
)

cross_grid
```
